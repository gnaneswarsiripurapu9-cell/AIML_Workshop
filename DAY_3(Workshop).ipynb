```python
final_data.drop(columns=['PassengerId','Name',
                         'Pclass','Sex','Ticket',
                         'Embarked'],inplace=True)
final_data.head()
```

```python
#Logistic Regression --> Binary Classification approach(survived -0/1)
#Wel will apply the modelling process

import numpy as pandas
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split,StratifiedKFold,StratifiedShuffleSplit,cross_val_score
from sklearn.metrics import (accuracy_score,precision_score,recall_score,confusion_matrix,classification_report,roc_curve,ConfusionMatrixDisplay)
```

```python
model_data = final_data.copy()
model_data
model_data.info()
```

```python
model_data.columns = model_data.columns.astype(str)
X = model_data.drop(columns='Survived')
y = model_data['Survived'].astype(int)
X.info()
```

```python
print("Input Shape:",X.shape)
print("Output Shape:",y.shape)
```

```python
#Holdoout Method -->splitting the data & Crossvalidation
X_train,X_test,y_train,y_test = train_test_split(X,y,
                                                 test_size=0.2,
                                                 stratify=y,
                                                 random_state=42)
logistic_model = Pipeline([
    ("scale",StandardScaler()),
    ("classifier",LogisticRegression(max_iter=1000,
                                     random_state=42))])
logistic_model
#start building the model
logistic_model.fit(X_train,y_train)
logistic_model_prob = logistic_model.predict_proba(X_test)[:,1]
logistic_predictions = logistic_model.predict(X_test)
print(logistic_model_prob[:10])
print(logistic_predictions[:10])
```

```python
from sklearn.metrics import f1_score,roc_auc_score
```

```python
y.shape
```

```python
#let's create an evaluate function
def evaluate_model(name,y_true,prediction,probability):
    """Evaluation Function for metrics"""
    tp,fp,fn,tn = confusion_matrix(y_true,prediction,labels=[0,1]).ravel()
    return {
        'Model_Name':name,
        'Accuracy':accuracy_score(y_true,prediction),
        'Precision':precision_score(y_true,prediction,zero_division=0),
        'Recall':recall_score(y_true,prediction,zero_division=0),
        'Specificity':tn / (tn + fp),
        'F1-Score':f1_score(y_true,prediction,zero_division=0),
        'ROC-AUC':roc_auc_score(y_true,probability),
        'TN':tn,"FP":fp,"TP":tp,"FN":fn,}
logistic_metrics = evaluate_model("Logistic Regression",
                                  y_test,
                                  logistic_predictions,
                                  logistic_model_prob)
logistic_metrics
```

```python
ConfusionMatrixDisplay.from_predictions(y_test,logistic_predictions,cmap='Blues',
                                        display_labels=['Didnot Survive','Survived'])
plt.title("Confusion Matrix")
plt.show()
```

```python
#We will check for the best models(Logisitic,Random Forest,Decision Tree) by
#performing validation and then will get the best model...
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
models = {
    "Logistic Regression":logistic_model,
    "Decision Tree":DecisionTreeClassifier(max_depth=5,
                                           min_samples_leaf=8,random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=150,
                                            min_samples_leaf=3,
                                            random_state=42,n_jobs=-1)}
test_outputs ={}
metric_rows = []
for name,model in models.items():
    model.fit(X_train,y_train)
    probability = model.predict_proba(X_test)[:,1]
    predictions = model.predict(X_test)
    test_outputs[name] = {
        'probability':probability,
        'Prediction':predictions}
    metric_rows.append(evaluate_model(name,y_test,predictions,probability))
holdout_metrics = pd.DataFrame(metric_rows)
holdout_metrics
```

```python
# Repeated random holdout (10 different stratified splits)
shuffle = StratifiedShuffleSplit(n_splits=10, test_size=0.20, random_state=42)
random_holdout_rows = []

for name, model in models.items():
    scores = []
    for train_index, test_index in shuffle.split(X, y):
        model.fit(X.iloc[train_index], y.iloc[train_index])
        probability = model.predict_proba(X.iloc[test_index])[:, 1]
        scores.append(roc_auc_score(y.iloc[test_index], probability))
    random_holdout_rows.append({
        "Model": name,
        "Mean ROC-AUC": np.mean(scores),
        "Std ROC-AUC": np.std(scores, ddof=1)
    })

random_holdout_results = pd.DataFrame(random_holdout_rows).set_index("Model")
print(random_holdout_results)
```

```python
# Stratified 5-fold cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_rows = []
cv_auc_scores = {}

for name, model in models.items():
    accuracy_scores = cross_val_score(model, X, y, cv=cv, scoring="accuracy", n_jobs=-1)
    f1_scores = cross_val_score(model, X, y, cv=cv, scoring="f1", n_jobs=-1)
    auc_scores = cross_val_score(model, X, y, cv=cv, scoring="roc_auc", n_jobs=-1)
    cv_auc_scores[name] = auc_scores
    cv_rows.append({
        "Model": name,
        "Accuracy mean": accuracy_scores.mean(),
        "F1 mean": f1_scores.mean(),
        "ROC-AUC mean": auc_scores.mean(),
        "ROC-AUC std": auc_scores.std(ddof=1),
    })

cv_results = pd.DataFrame(cv_rows).set_index("Model").sort_values("ROC-AUC mean", ascending=False)
print(cv_results)
```

```python
final_metrics = holdout_metrics[[
    "Accuracy", "Precision", "Recall", "Specificity", "ROC-AUC"
]].sort_values("ROC-AUC", ascending=False)

print("FINAL HOLDOUT METRICS")
display(final_metrics.round(3))

print("\nFINAL CROSS-VALIDATION METRICS")
display(cv_results.round(3))

print(f"\nBest model by mean CV ROC-AUC: {cv_results.index[0]}")
```

```python
# 1. Select the best model using cross-validation ROC-AUC

from sklearn.base import clone

best_model_name = cv_results["ROC-AUC mean"].idxmax()

# Create a fresh copy of the selected model
best_model = clone(models[best_model_name])

# Fit the best model using the complete final_data
best_model.fit(X, y)

print("Best model:", best_model_name)
print(
    "Mean CV ROC-AUC:",
    round(cv_results.loc[best_model_name, "ROC-AUC mean"], 3)
)
```

```python
# 3. Save the trained model as a Pickle file

import pickle
import sklearn

model_bundle = {
    "model": best_model,
    "model_name": best_model_name,
    "feature_names": X.columns.tolist(),
    "target_mapping": {
        0: "Did not survive",
        1: "Survived"
    },
    "selection_metric": "5-fold cross-validation ROC-AUC",
    "cv_roc_auc": float(
        cv_results.loc[best_model_name, "ROC-AUC mean"]
    ),
    "sklearn_version": sklearn.__version__
}

pickle_filename = "titanic_best_model.pkl"

with open(pickle_filename, "wb") as file:
    pickle.dump(model_bundle, file)

print("Model saved successfully:", pickle_filename)
```
