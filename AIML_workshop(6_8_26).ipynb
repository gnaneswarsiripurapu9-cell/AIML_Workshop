```python
data.isnull().sum()
```

```python
data['Age'] = data['Age'].fillna(data['Age'].mean(),inplace=True)
data.isnull().sum()
```

```python
data.drop(columns=['Cabin'],inplace=True)
```

```python
data.isnull().sum()
```

```python
data.columns
```

```python
Pclass = pd.get_dummies(data['Pclass'],
                        drop_first=True,
                        dtype=int)
Pclass
```

```python
gender = pd.get_dummies(data['Sex'],
                        drop_first=True,
                        dtype=int)
gender
```

```python
embarked = pd.get_dummies(data['Embarked'],drop_first=True,dtype=int)
embarked
```

```python
final_data = pd.concat([data,Pclass,gender,embarked],
                       axis=1)
final_data
```

```python
final_data.drop(columns=['PassengerId','Pclass','Name','Sex','Ticket','Embarked'],inplace=True)
final_data.head()
```
